```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
from pipeline.data import load_saved_folds
from pipeline.tokenization import load_fold_tokenizer
from pipeline.encoders import make_encoder
from pipeline.models import fit_model, predict_probabilities

data_dir = "../results/data/mnli-n50000-seed0"
subset, folds = load_saved_folds(data_dir)
tok = load_fold_tokenizer(data_dir, 0)
enc = make_encoder(tok, "overlap")

train_small = folds[0]["training"][:2000]
val_small = folds[0]["validation"][:500]
model, history = fit_model("mlp2", train_small, val_small, enc, "/tmp/mlp2_test",
                           smlp_args={"hidden_dim": 128, "num_epochs": 3})
probs = predict_probabilities("mlp2", [p for p, y in val_small], model, enc)
print(probs.shape, probs[0], probs[0].sum())
```
