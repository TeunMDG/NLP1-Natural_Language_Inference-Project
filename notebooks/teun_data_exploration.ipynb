```python
from collections import Counter

train_labels = mnli["train"]["label"]
print("train labels sizes: ", Counter(train_labels))

val_matched_labels = mnli["validation_matched"]["label"]
print("val matched labels sizes: ", Counter(val_matched_labels))

val_mismatch_labels = mnli["validation_mismatched"]["label"]
print("val mismatch labels sizes: ", Counter(val_mismatch_labels))
```

```python
train_genre = mnli["train"]["genre"]
print("train genres sizes: ", Counter(train_genre))

val_matched_genres = mnli["validation_matched"]["genre"]
print("val matched genre sizes: ", Counter(val_matched_genres))

val_mismatch_genres = mnli["validation_mismatched"]["genre"]
print("val mismatch genre sizes: ", Counter(val_mismatch_genres))
```
